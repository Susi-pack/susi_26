# Single scenario dashboard

```python
%matplotlib inline
from pathlib import Path

from susi.io.app_settings import AppSettings
import susi.io.load_output_data as load_output
from analysis.notebooks.components import (
    folder_selection,
    single_scenario_dashboard,
)
```

## Choose scenario folder

Browse to a custom folder below, or skip this cell to use the default projects root (`AppSettings().projects_root`) in the next one.

```python
folder_browser = folder_selection.build_folder_browser(AppSettings().projects_root)
```

```python
data_folder = folder_selection.resolve_start_folder(
    folder_browser, default=AppSettings().projects_root
)
project_dropdown = folder_selection.build_dropdown(data_folder, label="project")
```

```python
run_dropdown = folder_selection.build_run_dropdown(project_dropdown.value)
```

```python
stand_dropdown = folder_selection.build_dropdown(run_dropdown.value, label="stand")
```

```python
scenario_dropdown = folder_selection.build_dropdown(
    stand_dropdown.value, label="scenario"
)
```

```python
chosen_scenario_folder = scenario_dropdown.value
params = load_output.read_params_from_jsons(
    experiment_folderpath=chosen_scenario_folder
)
chosen_netcdf_filepath = Path(params.metadata["netcdf_output_filepath"])

report_data = single_scenario_dashboard.load_report_data(chosen_netcdf_filepath)
```
