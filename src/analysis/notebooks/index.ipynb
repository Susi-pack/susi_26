# SUSI results notebooks

A list of notebooks:

- [**Project summary**](project_summary.ipynb) -- aggregated summary table across all stands/scenarios in a project.
- [**Single**](single_netcdf.ipynb) -- inspect one scenario's metadata and quick-look plots of chosen variables.
- [**Compare scenarios for stand**](compare_scenarios_for_stand.ipynb) -- see which parameters differ (and which are shared) across a stand's scenarios.
- [**Single scenario dashboard**](single_scenario_dashboard.ipynb) -- 7 fixed figures (stand + hydrology + mass + carbon + nutrient balance) for one stand's scenario.
- [**Optimization**](optimization.ipynb) -- Pareto-optimal scenario combinations across a project's stands. Paroninkorpi only, for now (#216).
