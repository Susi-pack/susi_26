# Project summary

```python
from susi.io.app_settings import AppSettings
from analysis.notebooks.components import folder_selection, summary_table
```

## Choose project folder

```python
project_dropdown = folder_selection.build_dropdown(
    AppSettings().output_folder, label="project"
)
```

## Summary

```python
df_summary = summary_table.display_summary_table(project_dropdown.value)
```

## Export to CSV

```python
from susi.io.utils import get_project_root

export_dir = get_project_root() / "src" / "analysis" / "exports"
export_dir.mkdir(parents=True, exist_ok=True)

project_name = project_dropdown.value.name
export_path = export_dir / f"{project_name}_summary.csv"

if export_path.exists():
    print(f"Warning: overwriting existing file {export_path}")

df_summary.to_csv(export_path, index=False)
print(f"Saved to {export_path}")
```
